```python
file_path = "/Volumes/workspace/default/raw_data/crime_data_2024_to_present.csv" 
```

```python
raw_df = spark.read.csv(
    file_path, 
    header=True, 
    inferSchema=True
)

display(raw_df)
```

```python
(raw_df.write
  .format("delta")
  .mode("overwrite")
  .option("delta.columnMapping.mode", "name") 
  .saveAsTable("bronze_mydata"))

print("Success! Bronze Layer 'bronze_mydata' is now in your Catalog.")
```
